```python
import numpy as np
import pandas as pd
import scipy.stats as sps
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', 100)
pd.set_option('display.max_rows', 100)
```

```python
df = pd.read_csv(r'C:\Users\mechz\Documents\Jobs_Analysis\job_postings.csv')
display(df)
print(f'The mean, median, and standard deviation of the salaries are {np.mean(df['salary_year_avg'])}, {np.median(df['salary_year_avg'])}, and {np.std(df['salary_year_avg'])} respectively.')
```

```python
full_time_jobs_count = len(df[df['Full-time'] == 1]['salary_year_avg'])
print(f'The percentage of full-time jobs is {full_time_jobs_count / len(df):.3%}.')
na_entries_count = np.sum(df.isna())
print(f'There are {na_entries_count} undefined values.')
```

```python
health_insurance_salaries = df[df['job_health_insurance'] == 1]['salary_year_avg']
no_health_insurance_salaries = df[df['job_health_insurance'] == 0]['salary_year_avg']
health_insurance_mean = np.mean(health_insurance_salaries)
no_health_insurance_mean = np.mean(no_health_insurance_salaries)
health_insurance_count = len(health_insurance_salaries)
no_health_insurance_count = len(no_health_insurance_salaries)

test_results = sps.ttest_ind(health_insurance_salaries, no_health_insurance_salaries, equal_var=False)
print(f'The mean salary for jobs with health_insurance is {health_insurance_mean:.3f} while the mean salary for jobs is {no_health_insurance_mean:.3f}.')
print(f'The test statistic is {test_results.statistic:.3f}, the p-value is {test_results.pvalue:.6f}, and the degrees of freedom are {test_results.df:.3f}.')
print(f'The mean difference is {health_insurance_mean - no_health_insurance_mean:.3f}, the standard error is {test_results._standard_error:.3f}, and the 95% confidence interval is from {test_results.confidence_interval().low:.3f} to {test_results.confidence_interval().high:.3f}.')

fig, ax = plt.subplots(1, 2, figsize=(16, 8))
ax[0].bar(['Health Insurance', 'No Health Insurance'], [health_insurance_mean, no_health_insurance_mean])
ax[0].set_title('Average Salary')
ax[1].bar(['Health Insurance', 'No Health Insurance'], [health_insurance_count, no_health_insurance_count])
ax[1].set_title('Count')
plt.show()
```

```python
remote_salaries = df[df['job_work_from_home'] == 1]['salary_year_avg']
not_remote_salaries = df[df['job_work_from_home'] == 0]['salary_year_avg']
remote_mean = np.mean(remote_salaries)
not_remote_mean = np.mean(not_remote_salaries)
remote_count = len(remote_salaries)
not_remote_count = len(not_remote_salaries)

test_results = sps.ttest_ind(remote_salaries, not_remote_salaries, equal_var=False)
print(f'The mean salary for remote jobs is {remote_mean:.3f} while the mean salary for non-remote jobs is {not_remote_mean:.3f}.')
print(f'The test statistic is {test_results.statistic:.3f}, the p-value is {test_results.pvalue:.6f}, and the degrees of freedom are {test_results.df:.3f}.')
print(f'The mean difference is {remote_mean - not_remote_mean:.3f}, the standard error is {test_results._standard_error:.3f}, and the 95% confidence interval is from {test_results.confidence_interval().low:.3f} to {test_results.confidence_interval().high:.3f}.')

fig, ax = plt.subplots(1, 2, figsize=(16, 8))
ax[0].bar(['Remote', 'Not Remote'], [remote_mean, not_remote_mean])
ax[0].set_title('Average Salary')
ax[1].bar(['Remote', 'Not Remote'], [remote_count, not_remote_count])
ax[1].set_title('Count')
plt.show()
```

```python
no_degree_salaries = df[df['job_no_degree_mention'] == 1]['salary_year_avg']
degree_salaries = df[df['job_no_degree_mention'] == 0]['salary_year_avg']
no_degree_mean = np.mean(no_degree_salaries)
degree_mean = np.mean(degree_salaries)
no_degree_count = len(no_degree_salaries)
degree_count = len(degree_salaries)

test_results = sps.ttest_ind(no_degree_salaries, degree_salaries, equal_var=False)
print(f'The mean salary for jobs with no degree requirement is {no_degree_mean:.3f} while the mean salary for jobs with a degree requirement is {degree_mean:.3f}.')
print(f'The test statistic is {test_results.statistic:.3f}, the p-value is {test_results.pvalue:.6f}, and the degrees of freedom are {test_results.df:.3f}.')
print(f'The mean difference is {no_degree_mean - degree_mean:.3f}, the standard error is {test_results._standard_error:.3f}, and the 95% confidence interval is from {test_results.confidence_interval().low:.3f} to {test_results.confidence_interval().high:.3f}.')

fig, ax = plt.subplots(1, 2, figsize=(16, 8))
ax[0].bar(['No Degree Required', 'Degree Required'], [no_degree_mean, degree_mean])
ax[0].set_title('Average Salary')
ax[1].bar(['No Degree Required', 'Degree Required'], [no_degree_count, degree_count])
ax[1].set_title('Count')
plt.show()
```
